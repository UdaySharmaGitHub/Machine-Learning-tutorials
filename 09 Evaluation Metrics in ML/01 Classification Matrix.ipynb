# Classification Metrics: Logistic Regression and Decision Tree

In this notebook, we evaluate two **binary classification models** on the heart-disease dataset:

1. **Logistic Regression**
2. **Decision Tree Classifier**

A classification metric compares the actual class labels ($y$) with the labels predicted by a model ($\hat{y}$). These metrics help us understand not only how many predictions are correct, but also the types of errors the model makes.

The positive class is represented by `1` and the negative class by `0`. We use the following quantities throughout this notebook:

- **True Positive (TP):** The actual class is positive and the model predicts positive.
- **True Negative (TN):** The actual class is negative and the model predicts negative.
- **False Positive (FP):** The actual class is negative but the model predicts positive. This is a Type I error.
- **False Negative (FN):** The actual class is positive but the model predicts negative. This is a Type II error.

The models are trained on `X_train` and `y_train`, then evaluated on the unseen test data `X_test` and `y_test`.

```python
# Import pandas for loading and working with tabular data.
import pandas as pd

# The dataset is stored in the local datasets folder.
```

```python
# Load the heart-disease dataset into a pandas DataFrame.
df = pd.read_csv('datasets/heart.csv')
df
```

```python
# Import the helper used to divide the data into training and testing subsets.
from sklearn.model_selection import train_test_split

# Separate the input features from the target label.
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Keep 20% of the data for testing and use the remaining 80% for training.
# random_state makes the split reproducible.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)
```

```python
# Import two classification algorithms for comparison.
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

## Classification Metrics: Logistic Regression and Decision Tree

```python
# Create one Logistic Regression model and one Decision Tree model.
clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()
```

```python
# Train both models using only the training data.
clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
```

```python
# Generate predictions for the unseen test data.
y_pred_logistic_regression = clf1.predict(X_test)
y_pred_decision_tree = clf2.predict(X_test)
```

## 1. Accuracy

**Definition:** Accuracy is the proportion of all predictions that the model classifies correctly. It answers the question: *Out of every prediction, how many were correct?*

Correct predictions are the true positives and true negatives. The total number of predictions is $TP + TN + FP + FN$.

**Formula:**

$$
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
$$

**Derivation:**

1. Correct positive predictions = $TP$.
2. Correct negative predictions = $TN$.
3. Total correct predictions = $TP + TN$.
4. Total predictions = $TP + TN + FP + FN$.
5. Divide correct predictions by total predictions.

Accuracy is useful when the classes are reasonably balanced. With highly imbalanced classes, a model can have high accuracy while performing poorly on the minority class.

```python
# Import accuracy and confusion-matrix functions.
from sklearn.metrics import accuracy_score, confusion_matrix

# Accuracy measures the overall proportion of correct predictions.
print("Accuracy of Logistic Regression:", accuracy_score(y_test, y_pred_logistic_regression))
print("Accuracy of Decision Tree:", accuracy_score(y_test, y_pred_decision_tree))
```

## 2. Confusion Matrix

**Definition:** A confusion matrix is a $2 \\times 2$ table that counts the relationship between the actual class and the predicted class. It shows both correct predictions and the errors made by a classification model.

| Actual / Predicted | Predicted Negative | Predicted Positive |
|---|---:|---:|
| **Actual Negative** | $TN$ | $FP$ |
| **Actual Positive** | $FN$ | $TP$ |

The matrix is interpreted as follows:

- $TN$: correctly predicted negative observations.
- $FP$: negative observations incorrectly predicted as positive.
- $FN$: positive observations incorrectly predicted as negative.
- $TP$: correctly predicted positive observations.

### Type I and Type II Errors

Classification errors occur when the model's prediction does not match the actual class.

- **Type I Error (False Positive):** The actual class is negative, but the model predicts positive. In the confusion matrix, this is $FP$. For example, a healthy person is incorrectly classified as having heart disease. The model raises a false alarm.
- **Type II Error (False Negative):** The actual class is positive, but the model predicts negative. In the confusion matrix, this is $FN$. For example, a person with heart disease is incorrectly classified as healthy. The model fails to detect a positive case.

The error rates can be written as:

$$
\text{Type I Error Rate} = \frac{FP}{FP + TN}
$$

$$
\text{Type II Error Rate} = \frac{FN}{FN + TP}
$$

**Derivation of the error rates:**

1. For Type I error, consider all actually negative observations: $TN + FP$.
2. Among them, the incorrectly predicted positives are $FP$.
3. Therefore, Type I error rate is $FP/(TN + FP)$.
4. For Type II error, consider all actually positive observations: $TP + FN$.
5. Among them, the incorrectly predicted negatives are $FN$.
6. Therefore, Type II error rate is $FN/(TP + FN)$.

Type I and Type II errors usually involve a trade-off. Making a model more sensitive may reduce Type II errors but can increase Type I errors. The appropriate balance depends on the application. In medical screening, reducing Type II errors is often important because missing a disease can be more serious than raising a false alarm.

**Derivation of the matrix:**

1. Compare every actual label with its predicted label.
2. If actual and predicted labels are both negative, count one $TN$.
3. If actual is negative and predicted is positive, count one $FP$.
4. If actual is positive and predicted is negative, count one $FN$.
5. If actual and predicted labels are both positive, count one $TP$.
6. Place the four counts in their corresponding cells. Their sum equals the total number of observations:

$$
TP + TN + FP + FN = N
$$

The confusion matrix is the foundation for calculating accuracy, precision, recall, and F1-score.

```python
# Display the four confusion-matrix counts for each model:
# True Negative, False Positive, False Negative, and True Positive.
print("Confusion Matrix of Logistic Regression:\n", confusion_matrix(y_test, y_pred_logistic_regression))
print("Confusion Matrix of Decision Tree:\n", confusion_matrix(y_test, y_pred_decision_tree))
```

```python
# Convert the Logistic Regression confusion matrix into a labeled DataFrame.
# Rows represent actual classes and columns represent predicted classes.
logistic_confusion_matrix = pd.DataFrame(
    confusion_matrix(y_test, y_pred_logistic_regression),
    columns=['Predicted Negative', 'Predicted Positive'],
    index=['Actual Negative', 'Actual Positive']
)
logistic_confusion_matrix
```

```python
# Convert the Decision Tree confusion matrix into a labeled DataFrame.
decision_tree_confusion_matrix = pd.DataFrame(
    confusion_matrix(y_test, y_pred_decision_tree),
    columns=['Predicted Negative', 'Predicted Positive'],
    index=['Actual Negative', 'Actual Positive']
)
decision_tree_confusion_matrix
```

## 3. Precision, Recall, and F1-Score

These metrics focus on the positive class and are calculated from $TP$, $TN$, $FP$, and $FN$ in the confusion matrix.

--- 
### 3.1 Precision

**Definition:** Precision measures how trustworthy the model's positive predictions are. It answers: *When the model predicts positive, how often is it correct?*

Only predicted-positive observations are included in the denominator. These are $TP + FP$.

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

**Derivation:**

1. Correct positive predictions = $TP$.
2. All positive predictions = $TP + FP$.
3. Divide correct positive predictions by all positive predictions.

A low precision means the model produces many false positives.

--- 
### 3.2 Recall (Sensitivity)

**Definition:** Recall measures how many of the actual positive observations the model successfully identifies. It answers: *Of all truly positive observations, how many did the model find?*

All actual-positive observations are included in the denominator. These are $TP + FN$.

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

**Derivation:**

1. Correctly identified positive observations = $TP$.
2. All actual positive observations = $TP + FN$.
3. Divide the identified positives by all actual positives.

A low recall means the model misses many positive observations. In medical screening, recall is often especially important because a false negative may mean a disease is not detected.

--- 
### 3.3 F1-Score

**Definition:** The F1-score is the harmonic mean of precision and recall. It gives a single score that is high only when both precision and recall are high.

The harmonic mean of precision $P$ and recall $R$ is:

$$
F_1 = \frac{2PR}{P + R}
$$

Substitute the definitions of precision and recall:

$$
F_1 = \frac{2 \left(\frac{TP}{TP + FP}\right)\left(\frac{TP}{TP + FN}\right)}{\frac{TP}{TP + FP} + \frac{TP}{TP + FN}}
$$

Simplifying the expression gives the confusion-matrix form:

$$
F_1 = \frac{2TP}{2TP + FP + FN}
$$

**Derivation:**

1. Calculate precision using $TP/(TP + FP)$.
2. Calculate recall using $TP/(TP + FN)$.
3. Multiply precision and recall, then multiply by $2$.
4. Divide by the sum of precision and recall.
5. The resulting harmonic mean penalizes a large gap between precision and recall.

The F1-score is useful when false positives and false negatives both matter, especially when the classes are not balanced.

```python
# Import the metrics used to evaluate positive-class predictions.
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
# Calculate precision, recall, and F1-score for Logistic Regression.
# These metrics use the positive class as the class of interest.
print("For Logistic Regression Model")
print("-" * 50)
logistic_confusion_counts = pd.DataFrame(
    confusion_matrix(y_test, y_pred_logistic_regression),
    columns=['Predicted Negative', 'Predicted Positive'],
    index=['Actual Negative', 'Actual Positive']
)
print(logistic_confusion_counts)
print("-" * 50)
print("Precision =", precision_score(y_test, y_pred_logistic_regression))
print("Recall =", recall_score(y_test, y_pred_logistic_regression))
print("F1 score =", f1_score(y_test, y_pred_logistic_regression))
```

### ***for Desicion tree***

```python
# Calculate precision, recall, and F1-score for the Decision Tree.
print("For Decision Tree Model")
print("-" * 50)
decision_tree_confusion_counts = pd.DataFrame(
    confusion_matrix(y_test, y_pred_decision_tree),
    columns=['Predicted Negative', 'Predicted Positive'],
    index=['Actual Negative', 'Actual Positive']
)
print(decision_tree_confusion_counts)
print("-" * 50)
print("Precision =", precision_score(y_test, y_pred_decision_tree))
print("Recall =", recall_score(y_test, y_pred_decision_tree))
print("F1 score =", f1_score(y_test, y_pred_decision_tree))
```

```python
# Calculate each metric separately for class 0 and class 1.
# average=None returns one score per class instead of a single averaged score.
print("Precision by class - Logistic Regression:")
print(precision_score(y_test, y_pred_logistic_regression, average=None))

print("Precision by class - Decision Tree:")
print(precision_score(y_test, y_pred_decision_tree, average=None))

print("Recall by class - Logistic Regression:")
print(recall_score(y_test, y_pred_logistic_regression, average=None))

print("Recall by class - Decision Tree:")
print(recall_score(y_test, y_pred_decision_tree, average=None))

print("F1-score by class - Logistic Regression:")
print(f1_score(y_test, y_pred_logistic_regression, average=None))

print("F1-score by class - Decision Tree:")
print(f1_score(y_test, y_pred_decision_tree, average=None))
```

### How much Accuracy is Good ?

## 4. Summary and Interpretation

There is no universal accuracy value that is always considered good. The acceptable value depends on the problem, class balance, cost of errors, and the performance of a simple baseline model. Therefore, accuracy should be interpreted together with the confusion matrix, precision, recall, and F1-score.

| Metric | Formula | What it measures | A high value means | Important when |
|---|---|---|---|---|
| **Accuracy** | $\frac{TP + TN}{TP + TN + FP + FN}$ | Overall proportion of correct predictions | Most predictions are correct | Classes are reasonably balanced and both error types have similar cost |
| **Confusion Matrix** | Counts $TN$, $FP$, $FN$, and $TP$ | The complete pattern of correct predictions and errors | The model's errors are small or acceptable | We need to know which type of mistake the model makes |
| **Precision** | $\frac{TP}{TP + FP}$ | Reliability of positive predictions | Few predicted positives are false positives | False positives are costly |
| **Recall (Sensitivity)** | $\frac{TP}{TP + FN}$ | Coverage of actual positive cases | Few actual positives are missed | False negatives are costly, such as in medical screening |
| **F1-Score** | $\frac{2 \times \text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2TP}{2TP + FP + FN}$ | Balance between precision and recall | Precision and recall are both high | Classes are imbalanced or both error types matter |

### Final comparison of the two models

The code cells above calculate each metric for both **Logistic Regression** and the **Decision Tree**. Compare the models using the metric that best matches the application:

- Prefer higher **recall** when missing a positive case is more dangerous.
- Prefer higher **precision** when false alarms are more expensive.
- Prefer a higher **F1-score** when precision and recall should be balanced.
- Use **accuracy** as a general measure only after checking the class distribution and confusion matrix.

A model should not be selected from accuracy alone. The best model is the one whose error pattern and metric values fit the real-world objective.
