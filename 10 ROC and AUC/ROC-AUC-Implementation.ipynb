## 1. Load and inspect the dataset

Load the Pima Indians Diabetes dataset into a pandas DataFrame. Each row represents one patient, the clinical measurements are input features, and `Outcome` is the binary target (`1` = diabetes, `0` = no diabetes).

`data.head()` displays the first five rows so we can confirm the data loaded and inspect the columns.

**Expected output:** a five-row DataFrame preview with clinical feature columns and `Outcome`. This cell prepares the dataset; it does not calculate a model metric.

```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv')
data.head()
```

## 2. Separate features and target

Create the feature matrix $X$ by removing `Outcome`, and create the target vector $y$ from that column:

$$X = \text{data without Outcome}, \qquad y = \text{Outcome}$$

Each row of $X$ contains one patient's measurements; the corresponding entry in $y$ is that patient's class label. The assignments do not display output.

```python
X = data.drop('Outcome', axis=1)
y = data['Outcome']
```

## 3. Split data into training and test sets

`train_test_split` reserves 20% of the observations for final evaluation and uses 80% for training. `random_state=2` makes the random split repeatable.

$$\text{training proportion}=0.80, \qquad \text{test proportion}=0.20$$

`X_train` and `y_train` are used to fit the model; `X_test` and `y_test` are held out for computing scores and metrics. The cell only assigns variables, so it has no displayed output.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

## 4. Fit the Logistic Regression model

Fit a binary Logistic Regression model on the training features and labels. For a feature vector $x$, the model estimates the probability of the positive class using the sigmoid function:

$$P(y=1\mid x)=\sigma(w^Tx+b)=\frac{1}{1+e^{-(w^Tx+b)}}$$

The parameters are learned by minimizing binary cross-entropy (log loss):

$$
L=-\frac{1}{N}\sum_{i=1}^{N}\left[y_i\log(p_i)+(1-y_i)\log(1-p_i)\right]
$$

`max_iter=1000` allows extra optimization iterations if needed.

**Expected output:** the fitted estimator representation, such as `LogisticRegression(max_iter=1000)`. This confirms that fitting completed; it is not an accuracy or AUC score.

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

## 5. Get positive-class prediction scores

`predict_proba` returns a probability for each class. Since this is a binary problem, `[:, 1]` selects the probability of class `1` (diabetes) for every test observation:

$$s_i=P(y_i=1\mid x_i)$$

These continuous scores—not hard class labels—are used to sweep the decision threshold and construct the ROC curve.

**Expected output:** a NumPy array with one value in $[0,1]$ per test example. Each value is the model's estimated probability of diabetes.

```python
y_scores = model.predict_proba(X_test)[:, 1]
y_scores
```

## 6. Compute ROC points and thresholds

`roc_curve` does not select one optimal threshold. It evaluates a sequence of candidate score cutoffs and calculates the ROC point for each one:

1. Candidate thresholds are taken from the distinct model scores, with an infinite cutoff added for the no-positive-predictions point.
2. At each threshold $t$, the classification rule is:

$$\hat{y}_i(t)=\begin{cases}1,&s_i\ge t\\0,&s_i<t\end{cases}$$

   Here, $s_i$ is the score for observation $i$.
3. Using the resulting predictions, it calculates:

$$\text{TPR}(t)=\frac{TP}{TP+FN}, \qquad \text{FPR}(t)=\frac{FP}{FP+TN}$$

The function returns three aligned arrays: `thresholds`, `fpr`, and `tpr`. For each index $k$:

- `thresholds[k]` is the candidate cutoff.
- `(fpr[k], tpr[k])` is the ROC point produced by that cutoff.

The infinite threshold predicts no observations as positive, giving the ROC point $(0,0)$. This cell assigns the arrays without printing them.

```python
from sklearn.metrics import roc_curve, roc_auc_score

fpr,tpr,thresholds = roc_curve(y_test, y_scores)
```

## 7. Inspect the ROC arrays

Print the FPR, TPR, and threshold arrays returned by `roc_curve`. Values at the same array index belong to the same ROC operating point: `(fpr[i], tpr[i])` is produced by `thresholds[i]`.

**Expected output:** three labeled NumPy arrays separated by divider lines. FPR and TPR are rates between 0 and 1; thresholds are score cutoffs, generally listed from high to low. The arrays can be long because the ROC curve includes distinct score cutoffs.

```python
print("False Positive Rate:\n", fpr)
print("="*70)
print("True Positive Rate:\n", tpr)
print("="*70)
print("Thresholds:\n", thresholds)
```

## 8. Create the ROC curve

Plot the FPR values on the horizontal axis and TPR values on the vertical axis. Each ROC point corresponds to a threshold, and the diagonal from $(0,0)$ to $(1,1)$ is the random-ranking baseline.

The ROC coordinates use:

$$\text{TPR}=\frac{TP}{TP+FN}, \qquad \text{FPR}=\frac{FP}{FP+TN}$$

The blue line connects all computed ROC points. Markers label every tenth threshold to keep the graph readable; they are only a subset of the points on the line.

**Expected output:** an interactive Plotly ROC chart with axes from 0 to 1, the model curve, labeled threshold markers, and a dashed random baseline. A curve closer to the top-left indicates stronger discrimination. This cell visualizes the curve; the next plotting cell adds the AUC value.

```python
import plotly.graph_objects as go
import numpy as np


# Generate a trace for ROC curve
trace0 = go.Scatter(
    x=fpr,
    y=tpr,
    mode='lines',
    name='ROC curve'
)

# Only label every nth point to avoid cluttering
n = 10
indices = np.arange(len(thresholds)) % n == 0  # Choose indices where index mod n is 0

trace1 = go.Scatter(
    x=fpr[indices],
    y=tpr[indices],
    mode='markers+text',
    name='Threshold points',
    text=[f"Thr={thr:.2f}" for thr in thresholds[indices]],
    textposition='top center'
)


# Diagonal line
trace2 = go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Random (Area = 0.5)',
    line=dict(dash='dash')
)

data = [trace0, trace1, trace2]

# Define layout with square aspect ratio
layout = go.Layout(
    title='Receiver Operating Characteristic',
    xaxis=dict(title='False Positive Rate'),
    yaxis=dict(title='True Positive Rate'),
    autosize=False,
    width=800,
    height=800,
    showlegend=False
)

# Define figure and add data
fig = go.Figure(data=data, layout=layout)

# Show figure
fig.show()
```

## 9. Select a threshold with Youden's J

Youden's J chooses the threshold that maximizes the vertical distance above the random-guess diagonal:

$$J(t)=\text{TPR}(t)-\text{FPR}(t), \qquad t^*=\arg\max_t J(t)$$

In the arrays returned by `roc_curve`, calculate the selected threshold as:

$$i^*=\arg\max_i\left(\text{TPR}_i-\text{FPR}_i\right), \qquad t^*=\text{thresholds}[i^*]$$

The arrays are aligned by index, so the index with the largest Youden's $J$ value selects the matching threshold. With that threshold, predict positive when the model score $s_j\ge t^*$ and negative otherwise.

**Expected output:** one scalar threshold printed after `Optimal threshold is:`. This is a general-purpose criterion that weights sensitivity and specificity equally; it is not necessarily the best threshold when the application has explicit error costs or constraints.

```python
# Assume that fpr, tpr, thresholds have already been calculated
optimal_idx = np.argmax(tpr - fpr)
optimal_threshold = thresholds[optimal_idx]
print("Optimal threshold is:", optimal_threshold)
```

## 10. Plot the ROC curve and report its AUC

Recompute the ROC points, calculate the area under the curve, and display them in an interactive Plotly figure. AUC is computed from the ROC curve (equivalently, by summing trapezoid areas):

$$\text{AUC}=\int_0^1 \text{TPR}(u)\,du$$

where $u$ is FPR. The plot shows the ROC line, selected threshold markers, and the diagonal random baseline (AUC = 0.5). Marker labels are shown for every tenth threshold to reduce clutter; the line uses all ROC points.

**Expected output:** an interactive ROC chart. The legend reports the calculated AUC (formatted to two decimal places); points nearer the top-left are generally preferable. AUC summarizes ranking across thresholds and is not classification accuracy.

```python
import plotly.graph_objects as go
import numpy as np
from sklearn.metrics import roc_auc_score

# Assuming fpr, tpr, thresholds are already calculated as before
fpr, tpr, thresholds = roc_curve(y_test, y_scores)

# Calculate the AUC (Area Under the Curve)
roc_auc = roc_auc_score(y_test, y_scores)

# Generate a trace for ROC curve
trace0 = go.Scatter(
    x=fpr,
    y=tpr,
    mode='lines',
    name=f'ROC curve (Area = {roc_auc:.2f})'
)

# Only label every nth point to avoid cluttering
n = 10
indices = np.arange(len(thresholds)) % n == 0  # Choose indices where index mod n is 0

trace1 = go.Scatter(
    x=fpr[indices],
    y=tpr[indices],
    mode='markers+text',
    name='Threshold points',
    text=[f"Thr={thr:.2f}" for thr in thresholds[indices]],
    textposition='top center'
)

# Diagonal line
trace2 = go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Random (Area = 0.5)',
    line=dict(dash='dash')
)

data = [trace0, trace1, trace2]

# Define layout with square aspect ratio
layout = go.Layout(
    title='Receiver Operating Characteristic',
    xaxis=dict(title='False Positive Rate'),
    yaxis=dict(title='True Positive Rate'),
    autosize=False,
    width=800,
    height=800,
    showlegend=True
)

# Define figure and add data
fig = go.Figure(data=data, layout=layout)

# Show figure
fig.show()
```

## 11. Compare Logistic Regression and SVM

Train two classifiers on the same training/test split, compute their positive-class scores, and compare their ROC curves and AUC values. The SVM uses standardized features; the scaler is fitted only on training data and then applied to test data to avoid leaking test-set statistics:

$$z=\frac{x-\mu_{\text{train}}}{\sigma_{\text{train}}}$$

For each model, AUC is the area under its ROC curve. A larger AUC indicates better overall ranking on this test set; it does not identify a deployment threshold or imply that the model has that percentage accuracy.

**Expected output:** one interactive chart with Logistic Regression, SVM, and the random baseline. In the saved output, Logistic Regression has AUC **0.79** and SVM has AUC **0.81**; values may change if the data or software changes. The SVM's `probability=True` may also produce a scikit-learn deprecation warning in newer versions.

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.preprocessing import StandardScaler

# Assuming that X_train, X_test, y_train, y_test are already defined

# SVM requires feature scaling for better performance
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression model
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
lr_scores = lr_model.predict_proba(X_test)[:,1]

# SVM model
svm_model = SVC(probability=True)
svm_model.fit(X_train_scaled, y_train)
svm_scores = svm_model.predict_proba(X_test_scaled)[:,1]

# Generate ROC curve data for logistic regression model
lr_fpr, lr_tpr, lr_thresholds = roc_curve(y_test, lr_scores)
lr_auc = roc_auc_score(y_test, lr_scores)

# Generate ROC curve data for SVM model
svm_fpr, svm_tpr, svm_thresholds = roc_curve(y_test, svm_scores)
svm_auc = roc_auc_score(y_test, svm_scores)

# Generate a trace for the Logistic Regression ROC curve
trace0 = go.Scatter(
    x=lr_fpr,
    y=lr_tpr,
    mode='lines',
    name=f'Logistic Regression (Area = {lr_auc:.2f})'
)

# Generate a trace for the SVM ROC curve
trace1 = go.Scatter(
    x=svm_fpr,
    y=svm_tpr,
    mode='lines',
    name=f'SVM (Area = {svm_auc:.2f})'
)

# Diagonal line
trace2 = go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Random (Area = 0.5)',
    line=dict(dash='dash')
)

data = [trace0, trace1, trace2]

# Define layout with square aspect ratio
layout = go.Layout(
    title='Receiver Operating Characteristic',
    xaxis=dict(title='False Positive Rate'),
    yaxis=dict(title='True Positive Rate'),
    autosize=False,
    width=800,
    height=800,
    showlegend=True
)

# Define figure and add data
fig = go.Figure(data=data, layout=layout)

# Show figure
fig.show()
```
