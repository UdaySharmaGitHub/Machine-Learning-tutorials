# Mini-Batch Gradient Descent (MBGD) - Complete Implementation from Scratch

## Introduction

**Mini-Batch Gradient Descent** is an optimization algorithm that combines the benefits of both **Batch Gradient Descent (BGD)** and **Stochastic Gradient Descent (SGD)**. It divides the training dataset into small batches and performs parameter updates for each batch.

---

## Types of Gradient Descent - A Quick Comparison

| Type | Batch Size | Pros | Cons |
|------|------------|------|------|
| **Batch GD** | Entire Dataset | Stable convergence, accurate gradients | Slow for large datasets, high memory |
| **Stochastic GD** | 1 sample | Fast updates, can escape local minima | Noisy gradients, high variance |
| **Mini-Batch GD** | Small subset (e.g., 32, 64, 128) | Balance of speed & stability | Requires tuning batch size |

---

## Mathematical Foundation

### The Cost Function (Mean Squared Error)

For linear regression, we minimize the **Mean Squared Error (MSE)**:

$$J(\theta) = \frac{1}{m} \sum_{i=1}^{m} ( y^{(i)} - h_\theta(x^{(i)}))^2$$

Where:
- $m$ = number of training samples
- $h_\theta(x) = \theta^T x = w_1x_1 + w_2x_2 + ... + w_nx_n + b$ (hypothesis/prediction)
- $y^{(i)}$ = actual target value
- $\theta$ = parameters (weights and bias)

### Gradient Derivation

The gradient of the cost function with respect to parameters:

**For weights (coefficients):**
$$\frac{\partial J}{\partial w_j} = \frac{-2}{m} \sum_{i=1}^{m} (y^{(i)} - h_\theta(x^{(i)})) \cdot x_j^{(i)}$$

**For bias (intercept):**
$$\frac{\partial J}{\partial b} = \frac{-2}{m} \sum_{i=1}^{m} (y^{(i)} - h_\theta(x^{(i)}))$$

### Parameter Update Rule

$$\theta_{new} = \theta_{old} - \alpha \cdot \nabla J(\theta)$$

Where $\alpha$ is the learning rate.

---

## Mini-Batch Gradient Descent Algorithm

**For each epoch:**
1. Shuffle the training data
2. Divide data into mini-batches of size `batch_size`
3. For each mini-batch:
   - Compute predictions: $\hat{y} = Xw + b$
   - Compute gradients using only the mini-batch
   - Update parameters: $\theta = \theta - \alpha \cdot \nabla J(\theta)$

## 📌 Mini-Batch Gradient Descent (MBGD) – Detailed Explanation

Mini-batch Gradient Descent is a **hybrid optimization algorithm** that combines the stability of **Batch Gradient Descent** and the efficiency of **Stochastic Gradient Descent (SGD)**.

It is the **most commonly used optimization method in deep learning**.

---

# 1️⃣ Quick Recap: Where Mini-Batch Fits

| Type              | Data Used per Update       | Speed     | Stability           |
| ----------------- | -------------------------- | --------- | ------------------- |
| Batch GD          | Entire dataset             | Slow      | Very Stable         |
| SGD               | 1 sample                   | Very Fast | Noisy               |
| **Mini-Batch GD** | Small batch (e.g., 32, 64) | Fast      | Moderately Stable ✅ |

Mini-batch tries to get the **best of both worlds**.

---

# 2️⃣ Problem Setup

Assume:

* Dataset: $\{(x_1,y_1), (x_2,y_2), ..., (x_N,y_N)\}$
* Model parameters: $\theta$
* Loss function: $L(\theta)$

Total cost function:

$$J(\theta) = \frac{1}{N} \sum_{i=1}^{N} L_i(\theta)$$

Where:

* $N$ = total training samples
* $L_i(\theta)$ = loss for one data point

---

# 3️⃣ Mini-Batch Concept

Instead of computing gradient on:

* Entire dataset ❌
* One single sample ❌

We split data into **mini-batches** of size $B$:

$$B \ll N$$

Example:

* Total samples $N = 10,000$
* Batch size $B = 100$
* Number of batches per epoch = $\frac{10000}{100} = 100$

---

# 4️⃣ Mathematical Derivation

### Step 1: Mini-Batch Cost Function

For a mini-batch $B_k$:

$$J_{B_k}(\theta) = \frac{1}{B} \sum_{i \in B_k} L_i(\theta)$$

---

### Step 2: Gradient of Mini-Batch

$$\nabla J_{B_k}(\theta) = \frac{1}{B} \sum_{i \in B_k} \nabla L_i(\theta)$$

This is an **approximation** of the true gradient:

$$\nabla J(\theta) = \frac{1}{N} \sum_{i=1}^{N} \nabla L_i(\theta)$$

So:

$$\nabla J_{B_k}(\theta) \approx \nabla J(\theta)$$

---

### Step 3: Parameter Update Rule

$$\theta = \theta - \eta \nabla J_{B_k}(\theta)$$

Where:

* $\eta$ = learning rate

---

# 5️⃣ Algorithm Steps

### For each epoch:

1. Shuffle dataset
2. Split into mini-batches of size $B$
3. For each mini-batch:

   * Compute predictions
   * Compute loss
   * Compute gradient
   * Update parameters

---

# 6️⃣ Why Mini-Batch Works Better

### 🔹 1. Faster than Batch GD

* Does not require full dataset each update.
* Suitable for large datasets.

### 🔹 2. Less Noisy than SGD

* Gradient is averaged over multiple samples.
* More stable updates.

### 🔹 3. GPU Optimized

* Matrix operations on mini-batches use vectorization.
* Perfect for deep learning frameworks.

---

# 🔟 Variance Analysis (Important Concept)

Mini-batch gradient is:

$$\mathbb{E}[\nabla J_{B_k}(\theta)] = \nabla J(\theta)$$

It is an **unbiased estimator** of true gradient.

But it has variance:

$$Var(\nabla J_{B_k}) \propto \frac{1}{B}$$

👉 Larger batch → Lower variance
👉 Smaller batch → Higher variance

---

# 1️⃣1️⃣ Why It Generalizes Better

Small noise in updates:

* Helps escape shallow local minima
* Avoids sharp minima
* Leads to better generalization

This is why mini-batch is dominant in deep learning.

---

# 1️⃣2️⃣ Practical Example (Deep Learning)

In neural networks:

$$W = W - \eta \frac{1}{B} X^T ( \hat{y} - y )$$

Where:

* $X$ = mini-batch input matrix
* $\hat{y}$ = predictions
* $y$ = true labels

Everything is vectorized → very efficient.

## Custom Mini-Batch Gradient Descent Implementation from Scratch

---

### Vectorized Formulas

The key to efficient implementation is **vectorization** - using matrix operations instead of loops.

#### 1. Prediction (Forward Pass)

**Scalar Form:**
$$\hat{y}^{(i)} = w_1x_1^{(i)} + w_2x_2^{(i)} + ... + w_nx_n^{(i)} + b$$

**Vectorized Form:**
$$\hat{Y} = X \cdot W + b$$

Where:
- $X$ is the input matrix of shape `(batch_size, n_features)`
- $W$ is the weight vector of shape `(n_features,)`
- $b$ is the bias (scalar)
- $\hat{Y}$ is the prediction vector of shape `(batch_size,)`

**NumPy Implementation:**
```python
y_hat = np.dot(X_batch, self.coef_) + self.intercept_
```

---

#### 2. Gradient Computation

**For Intercept (Bias):**

$$\frac{\partial J}{\partial b} = -\frac{2}{m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)}) = -2 \cdot \text{mean}(y - \hat{y})$$

**NumPy Implementation:**
```python
intercept_der = -2 * np.mean(y_batch - y_hat)
```

---

**For Coefficients (Weights):**

$$\frac{\partial J}{\partial W} = -\frac{2}{m} X^T \cdot (Y - \hat{Y})$$

**Vectorized Explanation:**
- $(Y - \hat{Y})$ gives the error vector of shape `(batch_size,)`
- $X^T \cdot (Y - \hat{Y})$ performs dot product resulting in shape `(n_features,)`

**NumPy Implementation:**
```python
coef_der = -2 * np.dot((y_batch - y_hat), X_batch)
```

> **Note:** `np.dot((y - y_hat), X)` is equivalent to $X^T \cdot (y - \hat{y})$ when shapes align properly.

---

#### 3. Parameter Update

$$W_{new} = W_{old} - \alpha \cdot \frac{\partial J}{\partial W}$$

$$b_{new} = b_{old} - \alpha \cdot \frac{\partial J}{\partial b}$$

**NumPy Implementation:**
```python
self.coef_ = self.coef_ - (self.lr * coef_der)
self.intercept_ = self.intercept_ - (self.lr * intercept_der)
```

````python
import random

class MBGDRegressor:
    
    def __init__(self,batch_size,learning_rate=0.01, epochs=100):
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        self.weights = None
        self.bias = None
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(int(X_train.shape[0]/self.batch_size)):
                
                idx = random.sample(range(X_train.shape[0]),self.batch_size)
                
                # y_hat = w1*x1 + w2*x2 + w3*x3 + b
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
                #print("Shape of y_hat",y_hat.shape)
                
                # derivative of intercept and coefficients
                #intercept_der = -2 * np.sum(y_train[idx] - y_hat)
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                # intercept = intercept - (learning_rate * intercept_der)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)

                # coef_der = -2 * np.sum((y_train[idx] - y_hat) * X_train[idx])
                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                # coef = coef - (learning_rate * coef_der)
                self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
````

````python
mbgdregessor = MBGDRegressor(batch_size= int(X_train.shape[0]/10), learning_rate=0.01, epochs=50)
````

**Batch Size Calculation:**
```python
batch_size = int(X_train.shape[0] / 10)  # ~35 samples per batch
```

**Total Iterations per Epoch:** `n_samples / batch_size` ≈ 10 batches

````python
mbgdregessor.fit(X_train,y_train)
````

### Training Process Explained

During `fit()`:
1. **Initialize** weights to 1 and bias to 0
2. **For each epoch** (50 iterations):
   - Divide data into ~10 mini-batches
   - For each mini-batch:
     - Compute predictions using current weights
     - Calculate gradients (derivatives of loss)
     - Update weights and bias using gradient descent

The final `intercept_` and `coef_` values are the learned parameters.

````python
y_pred = mbgdregessor.predict(X_test)
````

## Comparing R² Score: Linear Regression vs Custom Mini-Batch GD

### Understanding R² Score (Coefficient of Determination)

$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}} = 1 - \frac{\sum(y_i - \hat{y}_i)^2}{\sum(y_i - \bar{y})^2}$$

Where:
- $SS_{res}$ = Residual sum of squares (prediction error)
- $SS_{tot}$ = Total sum of squares (variance in data)
- $R^2 = 1$ → Perfect fit
- $R^2 = 0$ → Model as good as predicting the mean

**Reference: Linear Regression R² Score ≈ 0.44**

````python
r2_score(y_test,y_pred)
````

## Mini-Batch Gradient Descent using Sklearn's SGDRegressor

### Why SGDRegressor for Mini-Batch?

Sklearn's `SGDRegressor` is designed for **Stochastic Gradient Descent** (single sample updates), but can be used for **Mini-Batch GD** by:

1. Using `partial_fit()` method for incremental learning
2. Passing mini-batches instead of single samples
3. Manually iterating over batches in epochs

### Key Parameters

| Parameter | Description | Value |
|-----------|-------------|-------|
| `learning_rate` | Learning rate schedule type | `'constant'` |
| `eta0` | Initial learning rate value | `0.1` |

### Implementation Strategy

```python
for epoch in range(n_epochs):
    idx = random.sample(range(n_samples), batch_size)
    sgd_regressor.partial_fit(X_train[idx], y_train[idx])
```

> **`partial_fit()`**: Performs a single gradient update step, perfect for online/incremental learning and mini-batch scenarios.

````python
from sklearn.linear_model import SGDRegressor
````

````python
sgd_regressor = SGDRegressor(learning_rate='constant', eta0=0.1)
````

````python
batch_size = int(X_train.shape[0]/10)

for i in range(100):
    
    idx = random.sample(range(X_train.shape[0]),batch_size)
    sgd_regressor.partial_fit(X_train[idx],y_train[idx])
````

````python
print("Implementing the MINI-BGD using Sklearn's SGDRegressor")
print("Coefficients: ",sgd_regressor.coef_)
print("Intercept: ",sgd_regressor.intercept_)
````

````python
y_pred = sgd_regressor.predict(X_test)
````

## Conclusion

### Summary of Mini-Batch Gradient Descent

Mini-Batch Gradient Descent provides an excellent balance between:
- **Computational efficiency** (smaller updates than BGD)
- **Convergence stability** (less noisy than SGD)
- **Memory efficiency** (doesn't require entire dataset in memory)

### Advantages of Mini-Batch GD

| Advantage | Description |
|-----------|-------------|
| **Vectorization** | Can leverage optimized linear algebra libraries (BLAS, GPU) |
| **Regularization Effect** | Mini-batch noise can help escape shallow local minima |
| **Scalability** | Works well with large datasets that don't fit in memory |
| **GPU Utilization** | Efficient batched operations on parallel hardware |

### Common Batch Sizes

| Batch Size | Use Case |
|------------|----------|
| 32 | Default for many deep learning tasks |
| 64-128 | Good balance for most problems |
| 256-512 | When using powerful GPUs |
| Full batch | Small datasets or when precision is critical |

### Tips for Tuning

1. **Start with batch_size = 32** and adjust based on memory/performance
2. **Use learning rate scheduling** (decay over time)
3. **Monitor convergence** - loss should decrease smoothly
4. **Consider momentum** for faster convergence
