## Why Categorical Naive Bayes?

All four features are **finite, unordered categories** — there is no natural numerical distance between *Sunny* and *Rain*. Gaussian NB assumes continuous features; Multinomial NB counts word occurrences. **Categorical NB** is designed exactly for this: each feature gets its own categorical distribution, estimated separately per class.

## Step 1 — The Naive Bayes Formula

Naive Bayes classifies a new example $\mathbf{x} = (x_1, x_2, \ldots, x_n)$ using **Bayes' Theorem**:

$$P(C_k \mid \mathbf{x}) = \frac{P(C_k) \cdot P(\mathbf{x} \mid C_k)}{P(\mathbf{x})}$$

The **"naive" independence assumption** lets us factorise the likelihood:

$$P(\mathbf{x} \mid C_k) = \prod_{i=1}^{n} P(x_i \mid C_k)$$

So the **full scoring formula** (ignoring the constant denominator $P(\mathbf{x})$) is:

$$\boxed{\text{score}(C_k, \mathbf{x}) = P(C_k) \cdot \prod_{i=1}^{n} P(x_i \mid C_k)}$$

The **decision rule** picks the class with the highest score:

$$\hat{y} = \arg\max_{C_k} \; P(C_k) \cdot \prod_{i=1}^{n} P(x_i \mid C_k)$$

### Working in Log-Space

Multiplying many small probabilities causes **numerical underflow** (the product rounds to 0). We convert to **log-space**, turning products into sums:

$$\hat{y} = \arg\max_{C_k} \left[ \log P(C_k) + \sum_{i=1}^{n} \log P(x_i \mid C_k) \right]$$

This is **mathematically identical** to the original formula — $\log$ is monotonically increasing, so the $\arg\max$ does not change.

## Step 2 — Prior Probability $P(C_k)$

The **prior** is the probability of each class *before* seeing any features — it is simply the class frequency in the training data:

$$\boxed{P(C_k) = \frac{N_k}{N}}$$

| Symbol | Meaning |
|---|---|
| $N_k$ | Number of training examples labelled $C_k$ |
| $N$ | Total number of training examples |

For the Play Tennis dataset:
- **Yes** appears in 9 of 14 days → $P(\text{Yes}) = 9/14 \approx 0.643$
- **No** appears in 5 of 14 days → $P(\text{No}) = 5/14 \approx 0.357$

```python
# Separate features and target
X = df.drop(columns=['day', 'play'])
y = df['play']

# Compute prior probabilities
N = len(y)                          # total examples
classes = y.unique()                # ['Yes', 'No']
class_counts = y.value_counts()     # N_k for each class

priors = class_counts / N

print('Class counts (N_k):')
print(class_counts.to_string())
print(f'\nTotal examples (N): {N}')
print('\nPrior probabilities P(C_k):')
for cls in classes:
    print(f'  P(play={cls}) = {class_counts[cls]}/{N} = {priors[cls]:.4f}')
```

## Step 3 — Likelihood $P(x_i = v \mid C_k)$ with Laplace Smoothing

The **likelihood** is the probability of seeing a specific feature value $v$ in class $C_k$.

### Raw (unsmoothed) estimate

$$P(x_i = v \mid C_k) = \frac{\text{count of } (x_i = v \text{ and class} = C_k)}{N_k}$$

**Problem:** If a value $v$ never appears in class $C_k$ in training, this gives $P = 0$, which collapses the entire product to **zero** — no matter how strong the other evidence. This is the **zero-frequency problem**.

### Laplace (Add-$\alpha$) Smoothing

We add a small pseudo-count $\alpha$ to every cell of the frequency table:

$$\boxed{\widehat{P}(x_i = v \mid C_k) = \frac{N_{kiv} + \alpha}{N_k + \alpha \cdot M_i}}$$

| Symbol | Meaning |
|---|---|
| $N_{kiv}$ | Count of training examples in class $C_k$ where feature $i$ equals $v$ |
| $N_k$ | Total training examples in class $C_k$ |
| $\alpha$ | Smoothing parameter ($\alpha = 1$: Laplace smoothing) |
| $M_i$ | Number of distinct values feature $i$ can take |

The denominator adds $\alpha \cdot M_i$ — one $\alpha$ per possible value — so all smoothed probabilities for feature $i$ still **sum to 1**.

```python
# Compute smoothed likelihoods P(x_i = v | C_k)
alpha = 1  # Laplace smoothing

# likelihoods[class][feature][value] = probability
likelihoods = {}

for cls in classes:
    likelihoods[cls] = {}
    X_cls = X[y == cls]               # rows belonging to this class
    N_k   = len(X_cls)                # N_k

    for feature in X.columns:
        M_i = X[feature].nunique()    # number of distinct values
        value_counts = X_cls[feature].value_counts()

        likelihoods[cls][feature] = {}
        for value in X[feature].unique():
            N_kiv = value_counts.get(value, 0)   # count (0 if unseen)
            prob  = (N_kiv + alpha) / (N_k + alpha * M_i)
            likelihoods[cls][feature][value] = prob

# Display the likelihood table for 'outlook'
print('Likelihoods for feature: outlook')
print(f'  (alpha={alpha}, Laplace smoothing)\n')
for cls in classes:
    N_k = class_counts[cls]
    M_i = X['outlook'].nunique()
    print(f'  Class = {cls}  (N_k={N_k}, M_i={M_i})')
    for val, prob in likelihoods[cls]['outlook'].items():
        count = X[y == cls]['outlook'].value_counts().get(val, 0)
        print(f'    P(outlook={val:<8} | play={cls}) = ({count}+{alpha}) / ({N_k}+{alpha}*{M_i}) = {prob:.4f}')
    print()
```

```python
# Display ALL likelihood tables as a neat summary
for feature in X.columns:
    print(f'=== P({feature} | play) ===')
    values = sorted(X[feature].unique())
    header = f"{'Value':<12}" + ''.join(f"{cls:<12}" for cls in classes)
    print(header)
    print('-' * (12 + 12 * len(classes)))
    for v in values:
        row = f"{v:<12}" + ''.join(f"{likelihoods[cls][feature][v]:<12.4f}" for cls in classes)
        print(row)
    print()
```

### Likelihood Heatmap

Each cell shows $\widehat{P}(\text{value} \mid \text{class})$. **Green** = high probability in that class; **Red** = low probability. Feature values where Yes and No colours differ strongly are the most **discriminative**.

```python
import matplotlib.pyplot as plt
import numpy as np

features = list(X.columns)

row_labels = []
matrix_no  = []
matrix_yes = []

for feat in features:
    for v in sorted(X[feat].unique()):
        row_labels.append(f'{feat}={v}')
        matrix_no.append(likelihoods['No'][feat][v])
        matrix_yes.append(likelihoods['Yes'][feat][v])

data = np.array([matrix_no, matrix_yes])

fig, ax = plt.subplots(figsize=(10, 4))
im = ax.imshow(data, aspect='auto', cmap='RdYlGn', vmin=0, vmax=1)
ax.set_xticks(range(len(row_labels)))
ax.set_xticklabels(row_labels, rotation=45, ha='right', fontsize=9)
ax.set_yticks([0, 1])
ax.set_yticklabels(['No', 'Yes'])
ax.set_title('Likelihood Heatmap  P(feature value | class)')
plt.colorbar(im, ax=ax, label='Probability')
for j in range(len(row_labels)):
    for i in range(2):
        val = data[i, j]
        color = 'black' if 0.25 < val < 0.75 else 'white'
        ax.text(j, i, f'{val:.2f}', ha='center', va='center', fontsize=8, color=color)
plt.tight_layout()
plt.show()
```

### The Zero-Frequency Problem — Demonstrated

Without Laplace smoothing ($\alpha = 0$), any feature value **absent** from a class during training gets $P = 0$. That single zero kills the entire product, making the class score zero regardless of all other evidence.

In this dataset: `Overcast` **never** appears in the `No` class — all 4 Overcast days are `Yes`. So without smoothing:

$$P(\text{Overcast} \mid \text{No}) = \frac{0}{5} = 0 \implies \text{score}(\text{No}) = 0 \quad \text{(always, no matter the other features)}$$

```python
# Build likelihoods WITHOUT smoothing (alpha=0) — raw MLE
likelihoods_raw = {}
for cls in classes:
    likelihoods_raw[cls] = {}
    X_cls = X[y == cls]
    N_k = len(X_cls)
    for feature in X.columns:
        vc = X_cls[feature].value_counts()
        likelihoods_raw[cls][feature] = {}
        for value in X[feature].unique():
            likelihoods_raw[cls][feature][value] = vc.get(value, 0) / N_k

print('Zero-frequency demonstration (alpha=0):')
print()
print(f'  P(outlook=Overcast | play=No)  = {likelihoods_raw["No"]["outlook"]["Overcast"]:.4f}  ← ZERO!')
print(f'  P(outlook=Overcast | play=Yes) = {likelihoods_raw["Yes"]["outlook"]["Overcast"]:.4f}')
print()

# What happens when we try to predict an Overcast day without smoothing?
test_day = {'outlook': 'Overcast', 'temp': 'Hot', 'humidity': 'Normal', 'wind': 'Weak'}

log_scores_raw = {}
for cls in classes:
    s = np.log(priors[cls])
    for feat, val in test_day.items():
        p = likelihoods_raw[cls][feat].get(val, 0)
        s += np.log(p) if p > 0 else float('-inf')
    log_scores_raw[cls] = s

print('  Log-scores for {Overcast, Hot, Normal, Weak} WITHOUT smoothing:')
for cls in classes:
    s = log_scores_raw[cls]
    label = '-inf  ← zero killed the product' if s == float('-inf') else f'{s:.4f}'
    print(f'    play={cls}: {label}')

print()
print('  WITH Laplace smoothing (alpha=1):')
# Inline prediction using already-computed smoothed likelihoods (no predict_proba needed yet)
log_scores_smoothed = {}
for cls in classes:
    s = np.log(priors[cls])
    for feat, val in test_day.items():
        p = likelihoods[cls][feat].get(
            val,
            alpha / (class_counts[cls] + alpha * X[feat].nunique())
        )
        s += np.log(p)
    log_scores_smoothed[cls] = s

max_log = max(log_scores_smoothed.values())
exp_s = {cls: np.exp(v - max_log) for cls, v in log_scores_smoothed.items()}
total = sum(exp_s.values())
post_smoothed = {cls: exp_s[cls] / total for cls in classes}

for cls in classes:
    print(f'    play={cls}: log-score={log_scores_smoothed[cls]:.4f},  P={post_smoothed[cls]:.4f}')

pred_smoothed = max(post_smoothed, key=post_smoothed.get)
print(f'\n  Prediction (with smoothing): play = {pred_smoothed}')
```

## Step 4 — The Prediction Algorithm

Given a new example $\mathbf{x} = (x_1, x_2, x_3, x_4)$, the prediction algorithm is:

**For each class $C_k$, compute the log-score:**

$$\text{log-score}(C_k) = \underbrace{\log P(C_k)}_{\text{log-prior}} + \underbrace{\sum_{i=1}^{4} \log P(x_i \mid C_k)}_{\text{sum of log-likelihoods}}$$

**Predict the class with the highest log-score:**

$$\hat{y} = \arg\max_{C_k} \; \text{log-score}(C_k)$$

**Convert log-scores to posterior probabilities** (for interpretability):

$$P(C_k \mid \mathbf{x}) = \frac{e^{\text{log-score}(C_k)}}{\displaystyle\sum_{j} e^{\text{log-score}(C_j)}}$$

This normalised form gives the model's **confidence** in each class.

```python
def predict_proba(sample: dict):
    """
    Returns (log_scores, posteriors) for each class.
    sample: dict mapping feature name -> feature value
    """
    log_scores = {}

    for cls in classes:
        # Start with the log-prior: log P(C_k)
        log_score = np.log(priors[cls])

        # Add log-likelihood for each feature: log P(x_i | C_k)
        for feature, value in sample.items():
            prob = likelihoods[cls][feature].get(
                value,
                alpha / (class_counts[cls] + alpha * X[feature].nunique())
            )
            log_score += np.log(prob)

        log_scores[cls] = log_score

    # Softmax: convert log-scores to probabilities
    max_log = max(log_scores.values())   # subtract max for numerical stability
    exp_scores = {cls: np.exp(s - max_log) for cls, s in log_scores.items()}
    total = sum(exp_scores.values())
    posteriors = {cls: exp_scores[cls] / total for cls in classes}

    return log_scores, posteriors


def predict(sample: dict) -> str:
    """Returns the predicted class label."""
    _, posteriors = predict_proba(sample)
    return max(posteriors, key=posteriors.get)
```

---

## Step 5 — Worked Prediction Example

Let us classify a new day with these conditions:

| Feature | Value |
|---|---|
| Outlook | Sunny |
| Temperature | Cool |
| Humidity | High |
| Wind | Strong |

We compute the **log-score for each class** step by step:

$$\text{score}(\text{Yes}) = \log P(\text{Yes}) + \log P(\text{Sunny}\mid\text{Yes}) + \log P(\text{Cool}\mid\text{Yes}) + \log P(\text{High}\mid\text{Yes}) + \log P(\text{Strong}\mid\text{Yes})$$

$$\text{score}(\text{No}) = \log P(\text{No}) + \log P(\text{Sunny}\mid\text{No}) + \log P(\text{Cool}\mid\text{No}) + \log P(\text{High}\mid\text{No}) + \log P(\text{Strong}\mid\text{No})$$

The class with the **higher score wins**.

```python
new_day = {
    'outlook' : 'Sunny',
    'temp'    : 'Cool',
    'humidity': 'High',
    'wind'    : 'Strong'
}

log_scores, posteriors = predict_proba(new_day)

print('=== Detailed Log-Score Breakdown ===\n')
for cls in classes:
    running = np.log(priors[cls])
    print(f'Class = {cls}')
    print(f'  log P({cls:<3})                    = log({priors[cls]:.4f}) = {running:.4f}')
    for feature, value in new_day.items():
        p = likelihoods[cls][feature][value]
        running += np.log(p)
        print(f'  log P({feature}={value:<8} | {cls}) = log({p:.4f}) = {np.log(p):.4f}')
    print(f'  {"─"*48}')
    print(f'  Log-score ({cls})                 = {log_scores[cls]:.4f}')
    print()

print('=== Posterior Probabilities (after normalisation) ===')
for cls in classes:
    print(f'  P(play={cls} | X) = {posteriors[cls]:.4f}  ({posteriors[cls]*100:.1f}%)')

prediction = predict(new_day)
print(f'\n>>> Prediction: play = {prediction}')
```

## Step 6 — Evaluate on the Full Training Set

We now run the classifier on all 14 examples and record the prediction alongside the actual label.

> **Note:** Because we evaluate on the same data used for training, this measures **training accuracy** — how well the model fits what it has seen. With only 14 examples a proper train/test split would be too small to be meaningful.

```python
results = []
for _, row in df.iterrows():
    sample    = row[['outlook', 'temp', 'humidity', 'wind']].to_dict()
    actual    = row['play']
    predicted = predict(sample)
    _, post   = predict_proba(sample)
    correct   = predicted == actual
    results.append({
        'Day'      : row['day'],
        'Actual'   : actual,
        'Predicted': predicted,
        'P(Yes)'   : f"{post['Yes']:.3f}",
        'P(No)'    : f"{post['No']:.3f}",
        'Correct'  : '✓' if correct else '✗'
    })

results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))

accuracy = sum(r['Correct'] == '✓' for r in results) / len(results)
print(f'\nTraining Accuracy: {accuracy:.4f}  ({accuracy*100:.1f}%)')
```

## Step 7 — Verify with scikit-learn `CategoricalNB`

scikit-learn's `CategoricalNB` implements the same Categorical Naive Bayes formula. We compare its predictions against our implementation to verify correctness.

> scikit-learn requires **integer-encoded** features (not strings), so we apply `OrdinalEncoder` first. This encoding is only for API compatibility — it does **not** imply any ordering between categories.

```python
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

# Encode features as integers (required by CategoricalNB)
enc_X = OrdinalEncoder()
enc_y = LabelEncoder()

X_enc = enc_X.fit_transform(X)
y_enc = enc_y.fit_transform(y)

# Train with same alpha
sk_model = CategoricalNB(alpha=alpha)
sk_model.fit(X_enc, y_enc)

# Predictions
sk_preds_enc = sk_model.predict(X_enc)
sk_preds     = enc_y.inverse_transform(sk_preds_enc)
our_preds    = [
    predict(row[['outlook', 'temp', 'humidity', 'wind']].to_dict())
    for _, row in df.iterrows()
]

# Compare
comparison = pd.DataFrame({
    'Day'    : df['day'].values,
    'Actual' : y.values,
    'Ours'   : our_preds,
    'sklearn': sk_preds,
    'Match'  : ['✓' if a == b else '✗' for a, b in zip(our_preds, sk_preds)]
})

print(comparison.to_string(index=False))
print(f'\nAll predictions match sklearn: {all(comparison["Match"] == "✓")}')
print(f'Our accuracy    : {sum(p == a for p, a in zip(our_preds, y.values)) / len(y):.4f}')
print(f'sklearn accuracy: {sk_model.score(X_enc, y_enc):.4f}')
```

## Step 10 — Feature Importance via Log-Likelihood Ratio

The **Log-Likelihood Ratio (LLR)** measures how much each feature value shifts the decision toward `Yes` vs `No`:

$$\text{LLR}(v) = \log\frac{P(v \mid \text{Yes})}{P(v \mid \text{No})}$$

| LLR | Interpretation |
|---|---|
| **Positive** | value is evidence *for* Yes |
| **Negative** | value is evidence *for* No |
| **Zero** | value provides no discriminative signal |

```python
import matplotlib.pyplot as plt

llr_data = []
for feat in X.columns:
    for v in sorted(X[feat].unique()):
        p_yes = likelihoods['Yes'][feat][v]
        p_no  = likelihoods['No'][feat][v]
        llr   = np.log(p_yes / p_no)
        llr_data.append({'Feature=Value': f'{feat}={v}', 'LLR': llr})

llr_data.sort(key=lambda x: x['LLR'])
labels = [d['Feature=Value'] for d in llr_data]
values = [d['LLR'] for d in llr_data]
colors = ['#22c55e' if v > 0 else '#ef4444' for v in values]

print(f'  {"Feature=Value":<22}  {"LLR":>8}  Direction')
print('  ' + '─' * 44)
for lbl, val in zip(labels, values):
    direction = '→ Yes' if val > 0 else '→ No '
    print(f'  {lbl:<22}  {val:>8.4f}  {direction}')

fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.barh(labels, values, color=colors, edgecolor='white')
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Log-Likelihood Ratio  log P(v|Yes) / P(v|No)')
ax.set_title('Feature Importance — which values favour Yes (green) or No (red)?')
for bar, val in zip(bars, values):
    offset = 0.02 if val >= 0 else -0.02
    ha = 'left' if val >= 0 else 'right'
    ax.text(val + offset, bar.get_y() + bar.get_height() / 2,
            f'{val:.2f}', va='center', ha=ha, fontsize=8)
plt.tight_layout()
plt.show()
```

## Step 11 — Alpha Sensitivity

How sensitive is training accuracy to the smoothing parameter $\alpha$? We sweep $\alpha$ from $0.01$ to $10$ and recompute accuracy each time.

> With only 14 examples, training accuracy is a rough proxy — but the plot shows whether heavy smoothing ($\alpha \gg 1$) noticeably degrades the fit by pulling all likelihoods toward the uniform distribution.

```python
import matplotlib.pyplot as plt

alpha_values = [0.01, 0.05, 0.1, 0.2, 0.5, 1, 2, 5, 10]
accuracies   = []

for a in alpha_values:
    lh = {}
    for cls in classes:
        lh[cls] = {}
        X_cls = X[y == cls]
        N_k = len(X_cls)
        for feat in X.columns:
            M_i = X[feat].nunique()
            vc  = X_cls[feat].value_counts()
            lh[cls][feat] = {
                v: (vc.get(v, 0) + a) / (N_k + a * M_i)
                for v in X[feat].unique()
            }

    def predict_a(sample):
        scores = {}
        for cls in classes:
            s = np.log(priors[cls])
            for feat, val in sample.items():
                p = lh[cls][feat].get(val, a / (class_counts[cls] + a * X[feat].nunique()))
                s += np.log(p)
            scores[cls] = s
        return max(scores, key=scores.get)

    correct = sum(
        predict_a(row[['outlook', 'temp', 'humidity', 'wind']].to_dict()) == row['play']
        for _, row in df.iterrows()
    )
    accuracies.append(correct / len(df))

print(f'{"alpha":<8} {"accuracy":>10}')
print('-' * 20)
for a, acc in zip(alpha_values, accuracies):
    marker = ' ← default' if a == 1 else ''
    print(f'{a:<8} {acc:>10.4f}  ({acc*100:.1f}%){marker}')

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(alpha_values, accuracies, marker='o', color='#3b82f6', linewidth=2)
ax.set_xscale('log')
ax.set_ylim(0, 1.05)
ax.set_xlabel('Alpha (log scale)')
ax.set_ylabel('Training Accuracy')
ax.set_title('Alpha Sensitivity — Training Accuracy vs Smoothing Strength')
ax.axvline(1, color='gray', linestyle='--', linewidth=1, label='alpha=1 (Laplace)')
ax.legend()
plt.tight_layout()
plt.show()
```

## Step 12 — Decision Table for All 36 Weather Combinations

There are $3 \times 3 \times 2 \times 2 = 36$ distinct combinations of the four features. Running the classifier on every combination gives a complete **decision table** — we can see the full decision boundary of the model.

```python
from itertools import product

outlooks   = sorted(X['outlook'].unique())
temps      = sorted(X['temp'].unique())
humidities = sorted(X['humidity'].unique())
winds      = sorted(X['wind'].unique())

rows = []
for o, t, h, w in product(outlooks, temps, humidities, winds):
    sample = {'outlook': o, 'temp': t, 'humidity': h, 'wind': w}
    _, post = predict_proba(sample)
    pred    = predict(sample)
    rows.append({
        'Outlook' : o,
        'Temp'    : t,
        'Humidity': h,
        'Wind'    : w,
        'Pred'    : pred,
        'P(Yes)'  : f"{post['Yes']:.3f}",
        'P(No)'   : f"{post['No']:.3f}",
    })

combos_df = pd.DataFrame(rows)
print(f'Total combinations : {len(combos_df)}')
print(f'Predicted Yes      : {(combos_df["Pred"] == "Yes").sum()}')
print(f'Predicted No       : {(combos_df["Pred"] == "No").sum()}')
print()
print(combos_df.to_string(index=False))
```

## Step 13 — Leave-One-Out Cross-Validation (LOOCV)

**LOOCV** gives a fairer estimate of generalisation than training accuracy. On each of the 14 folds:

1. Hold out **one example** as the test set
2. **Re-train** from scratch on the remaining 13 examples (re-estimate priors and likelihoods)
3. **Predict** the held-out example using the fold's parameters

The **LOOCV accuracy** tells us how well the model performs on *unseen* data. The gap between training and LOOCV accuracy is the **optimism** — how much the model overfits.

```python
loocv_results = []

for test_idx in range(len(df)):
    train_df = df.drop(index=test_idx).reset_index(drop=True)
    test_row  = df.iloc[test_idx]

    X_train = train_df.drop(columns=['day', 'play'])
    y_train = train_df['play']

    N_tr   = len(y_train)
    cc_tr  = y_train.value_counts()
    pr_tr  = cc_tr / N_tr
    cls_tr = y_train.unique()

    lh_tr = {}
    for cls in cls_tr:
        lh_tr[cls] = {}
        X_cls = X_train[y_train == cls]
        N_k = len(X_cls)
        for feat in X_train.columns:
            M_i = X_train[feat].nunique()
            vc  = X_cls[feat].value_counts()
            lh_tr[cls][feat] = {
                v: (vc.get(v, 0) + alpha) / (N_k + alpha * M_i)
                for v in X_train[feat].unique()
            }

    sample = test_row[['outlook', 'temp', 'humidity', 'wind']].to_dict()
    scores_tr = {}
    for cls in cls_tr:
        s = np.log(pr_tr[cls])
        for feat, val in sample.items():
            p = lh_tr[cls][feat].get(
                val,
                alpha / (cc_tr[cls] + alpha * X_train[feat].nunique())
            )
            s += np.log(p)
        scores_tr[cls] = s

    pred = max(scores_tr, key=scores_tr.get)
    loocv_results.append({
        'Day'      : test_row['day'],
        'Actual'   : test_row['play'],
        'Predicted': pred,
        'Correct'  : '✓' if pred == test_row['play'] else '✗'
    })

loocv_df = pd.DataFrame(loocv_results)
print(loocv_df.to_string(index=False))

loocv_acc = sum(r['Correct'] == '✓' for r in loocv_results) / len(loocv_results)
train_acc = sum(
    predict(row[['outlook', 'temp', 'humidity', 'wind']].to_dict()) == row['play']
    for _, row in df.iterrows()
) / len(df)

print(f'\nTraining accuracy : {train_acc:.4f}  ({train_acc*100:.1f}%)')
print(f'LOOCV accuracy    : {loocv_acc:.4f}  ({loocv_acc*100:.1f}%)')
print(f'Optimism gap      : {train_acc - loocv_acc:.4f}  ({(train_acc - loocv_acc)*100:.1f} pp)')
```

- **Laplace smoothing** ($\alpha = 1$) prevents zero probabilities for unseen feature values.
- **Log-space computation** prevents numerical underflow when multiplying many small probabilities.

- The **LOOCV accuracy** gives a fairer generalisation estimate than training accuracy.
