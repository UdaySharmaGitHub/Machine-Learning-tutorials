```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from sklearn.datasets import load_diabetes
data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
X = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=45,test_size=0.2)
```

### Implementing the Linear Regression to calculate R2_score & RMSE and Coefficient and Intercept

```python
from sklearn.linear_model import LinearRegression
linear_regressor_model = LinearRegression()
```

```python
linear_regressor_model.fit(X_train,y_train)
```

```python
print("Finding the Coefficients/weights and intercepts of the Linear Regression Model")
print("Coefficients/weights:", linear_regressor_model.coef_)
print("Intercept:", linear_regressor_model.intercept_)
```

```python
y_pred = linear_regressor_model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error

print("R2 Score:", r2_score(y_test,y_pred))
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y_test,y_pred)))
```

### Implementing the Ridge Regression or L2 Regularization to add penalize the Cost function to get avoid the overfitting the model

```python
from sklearn.linear_model import Ridge
Ridge_regressor_model = Ridge(alpha=100000)
```

```python
Ridge_regressor_model.fit(X_train,y_train)
```

```python
print("Implementing the Ridge Regression Model with alpha=100000")
print("Finding the Coefficients/weights and intercepts of the Ridge Regression Model")
print("Coefficients/weights:", Ridge_regressor_model.coef_)
print("Intercept:", Ridge_regressor_model.intercept_)
```

```python
y_pred= Ridge_regressor_model.predict(X_test) 
y_pred
```

```python
print(f'r2_score for Ridge Regression Model with alpha=100000: {r2_score(y_test,y_pred)}')
print(f'Root Mean Squared Error for Ridge Regression Model with alpha=100000: {np.sqrt(mean_squared_error(y_test,y_pred))}')
```

### Demonstrating Underfitting, Fitting and Overfitting with Different Alpha Values on Diabetes Dataset

The **alpha** parameter in Ridge Regression controls the strength of regularization:
- **Very High Alpha (e.g., 10000+)**: Strong regularization → Coefficients shrink towards zero → **Underfitting**
- **Optimal Alpha**: Balanced regularization → Good generalization → **Properly Fitted**
- **Very Low Alpha (e.g., 0)**: No regularization → Same as Linear Regression → Risk of **Overfitting**

```python
# Testing different alpha values to demonstrate underfitting, fitting, and overfitting
alphas = [0, 0.1, 1, 10, 100, 1000, 10000, 100000, 1000000]

# Store results
results = []

for alpha in alphas:
    if alpha == 0:
        # Alpha=0 is equivalent to Linear Regression (no regularization)
        model = LinearRegression()
    else:
        model = Ridge(alpha=alpha)
    
    model.fit(X_train, y_train)
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)
    
    train_r2 = r2_score(y_train, y_pred_train)
    test_r2 = r2_score(y_test, y_pred_test)
    train_rmse = np.sqrt(mean_squared_error(y_train, y_pred_train))
    test_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
    
    results.append({
        'Alpha': alpha,
        'Train R2': train_r2,
        'Test R2': test_r2,
        'Train RMSE': train_rmse,
        'Test RMSE': test_rmse,
        'Coefficients Sum': np.sum(np.abs(model.coef_))
    })

# Create DataFrame to display results
results_df = pd.DataFrame(results)
print("Effect of Alpha on Ridge Regression Performance:")
print("="*80)
print(results_df.to_string(index=False))
print("="*80)
```

```python
# Visualize R2 Score vs Alpha - Demonstrating Underfitting, Fitting, Overfitting
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot R2 Scores
axes[0].semilogx(results_df['Alpha'].replace(0, 0.01), results_df['Train R2'], 'b-o', label='Train R2', markersize=8)
axes[0].semilogx(results_df['Alpha'].replace(0, 0.01), results_df['Test R2'], 'r-s', label='Test R2', markersize=8)
axes[0].set_xlabel('Alpha (log scale)', fontsize=12)
axes[0].set_ylabel('R2 Score', fontsize=12)
axes[0].set_title('R2 Score vs Alpha\n(Higher is Better)', fontsize=14)
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)
axes[0].axvspan(0.01, 1, alpha=0.2, color='red', label='Overfitting Zone')
axes[0].axvspan(1, 100, alpha=0.2, color='green', label='Optimal Zone')
axes[0].axvspan(10000, 1000000, alpha=0.2, color='blue', label='Underfitting Zone')

# Plot RMSE
axes[1].semilogx(results_df['Alpha'].replace(0, 0.01), results_df['Train RMSE'], 'b-o', label='Train RMSE', markersize=8)
axes[1].semilogx(results_df['Alpha'].replace(0, 0.01), results_df['Test RMSE'], 'r-s', label='Test RMSE', markersize=8)
axes[1].set_xlabel('Alpha (log scale)', fontsize=12)
axes[1].set_ylabel('RMSE', fontsize=12)
axes[1].set_title('RMSE vs Alpha\n(Lower is Better)', fontsize=14)
axes[1].legend(fontsize=10)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nInterpretation:")
print("- RED zone (Low Alpha): Risk of Overfitting - Model fits training data too well but may not generalize")
print("- GREEN zone (Optimal Alpha): Properly Fitted - Good balance between bias and variance")
print("- BLUE zone (High Alpha): Underfitting - Model is too simple, high bias")
```

```python
# Visualize how coefficients change with different alpha values
feature_names = data.feature_names
alphas_to_plot = [0, 1, 10, 100, 1000, 10000, 100000]
coef_matrix = []

for alpha in alphas_to_plot:
    if alpha == 0:
        model = LinearRegression()
    else:
        model = Ridge(alpha=alpha)
    model.fit(X_train, y_train)
    coef_matrix.append(model.coef_)

coef_matrix = np.array(coef_matrix)

plt.figure(figsize=(12, 6))
for i, feature in enumerate(feature_names):
    plt.semilogx([a if a > 0 else 0.1 for a in alphas_to_plot], coef_matrix[:, i], '-o', label=feature, markersize=6)

plt.xlabel('Alpha (log scale)', fontsize=12)
plt.ylabel('Coefficient Value', fontsize=12)
plt.title('Ridge Regression: Coefficient Shrinkage with Increasing Alpha\n(Coefficients approach zero as alpha increases)', fontsize=14)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("\nKey Observation:")
print("As alpha increases, coefficients shrink towards zero, reducing model complexity")
print("- Low alpha: Large coefficients → Complex model (overfitting risk)")
print("- High alpha: Small coefficients → Simple model (underfitting risk)")
```

```python
# Detailed comparison of Underfitting, Properly Fitted, and Overfitting scenarios
print("="*80)
print("DETAILED COMPARISON: UNDERFITTING vs PROPERLY FITTED vs OVERFITTING")
print("="*80)

scenarios = {
    'OVERFITTING (alpha=0)': 0,
    'PROPERLY FITTED (alpha=10)': 10,
    'UNDERFITTING (alpha=100000)': 100000
}

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for idx, (scenario_name, alpha) in enumerate(scenarios.items()):
    if alpha == 0:
        model = LinearRegression()
    else:
        model = Ridge(alpha=alpha)
    
    model.fit(X_train, y_train)
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)
    
    train_r2 = r2_score(y_train, y_pred_train)
    test_r2 = r2_score(y_test, y_pred_test)
    
    print(f"\n{scenario_name}")
    print(f"  Train R2: {train_r2:.4f} | Test R2: {test_r2:.4f}")
    print(f"  Gap (Train-Test): {train_r2 - test_r2:.4f}")
    print(f"  Sum of |Coefficients|: {np.sum(np.abs(model.coef_)):.4f}")
    
    # Plot actual vs predicted for test set
    axes[idx].scatter(y_test, y_pred_test, alpha=0.6, edgecolors='k', linewidth=0.5)
    axes[idx].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Perfect Fit')
    axes[idx].set_xlabel('Actual Values', fontsize=10)
    axes[idx].set_ylabel('Predicted Values', fontsize=10)
    axes[idx].set_title(f'{scenario_name}\nTest R2: {test_r2:.4f}', fontsize=11)
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("SUMMARY:")
print("- OVERFITTING: Small gap but may not generalize to new data")
print("- PROPERLY FITTED: Best test performance with balanced coefficients")
print("- UNDERFITTING: Large gap, model too constrained, cannot capture patterns")
print("="*80)
```

```python
# Using Cross-Validation to find the optimal alpha (avoiding underfitting and overfitting)
from sklearn.linear_model import RidgeCV

# Define range of alphas to test
alphas_cv = np.logspace(-3, 6, 100)  # From 0.001 to 1,000,000

# RidgeCV automatically performs cross-validation to find optimal alpha
ridge_cv = RidgeCV(alphas=alphas_cv, cv=5, scoring='r2')
ridge_cv.fit(X_train, y_train)

print("Finding Optimal Alpha using 5-Fold Cross-Validation")
print("="*60)
print(f"Optimal Alpha: {ridge_cv.alpha_:.4f}")
print(f"Train R2 Score: {ridge_cv.score(X_train, y_train):.4f}")
print(f"Test R2 Score: {ridge_cv.score(X_test, y_test):.4f}")
print("="*60)

# Compare with extreme cases
print("\nComparison with Extreme Alpha Values:")
print("-"*60)
for alpha, label in [(0.001, 'Very Low (Overfitting risk)'), 
                      (ridge_cv.alpha_, 'Optimal (CV selected)'), 
                      (100000, 'Very High (Underfitting risk)')]:
    model = Ridge(alpha=alpha)
    model.fit(X_train, y_train)
    print(f"Alpha={alpha:>10.4f} ({label:>25}): Test R2 = {model.score(X_test, y_test):.4f}")
```

### Summary: How Alpha Affects Model Fitting on Diabetes Dataset

| Alpha Value | Effect | Model State | Characteristics |
|-------------|--------|-------------|-----------------|
| **Very Low (0-0.1)** | Minimal regularization | **Overfitting** | Large coefficients, good train score, may fail on new data |
| **Optimal (1-100)** | Balanced regularization | **Properly Fitted** | Moderate coefficients, good generalization |
| **Very High (10000+)** | Strong regularization | **Underfitting** | Coefficients near zero, poor performance on both train & test |

**Key Takeaways:**
1. Alpha controls the trade-off between bias and variance
2. Low alpha → High variance (overfitting)
3. High alpha → High bias (underfitting)
4. Use cross-validation (RidgeCV) to find optimal alpha automatically

### Now showing the underfitting , Fitted and Overfitting on the randome data 

```python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```
