```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y = make_regression(n_samples = 100, n_features=1 , n_informative= 1, noise=20, random_state=13)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_regressor_model = LinearRegression()
linear_regressor_model.fit(X,y)
print("Finding the Coefficients/weights and intercepts of the Linear Regression Model")
print("Coefficients:", linear_regressor_model.coef_)
print("Intercept:", linear_regressor_model.intercept_)
```

## Implementing the Ridge Regression 

```python
from sklearn.linear_model import Ridge
```

```python
ridge_regressor_model_1 = Ridge(alpha=1)
ridge_regressor_model_1.fit(X,y)
print("Implementing the Ridge Regression Model with alpha=1")
print("Finding the Coefficients/weights and intercepts of the Ridge Regression Model")
print("Coefficients/weights:", ridge_regressor_model_1.coef_)
print("Intercept:", ridge_regressor_model_1.intercept_)
```

```python
ridge_regressor_model_2 = Ridge(alpha=10)
ridge_regressor_model_2.fit(X,y)
print("Implementing the Ridge Regression Model with alpha=10")
print("Finding the Coefficients/weights and intercepts of the Ridge Regression Model")
print("Coefficients/weights:", ridge_regressor_model_2.coef_)
print("Intercept:", ridge_regressor_model_2.intercept_)
```

```python
ridge_regressor_model_3 = Ridge(alpha=100)
ridge_regressor_model_3.fit(X,y)
print("Implementing the Ridge Regression Model with alpha=100")
print("Finding the Coefficients/weights and intercepts of the Ridge Regression Model")
print("Coefficients/weights:", ridge_regressor_model_3.coef_)
print("Intercept:", ridge_regressor_model_3.intercept_)
```

```python
plt.plot(X,y,'b.') # first plot the data points
plt.plot(X,linear_regressor_model.predict(X),color='red',label='alpha=0') # then plot the linear regression line
plt.plot(X,ridge_regressor_model_1.predict(X),color='green',label='alpha=1') # then plot the ridge regression line with alpha=1
plt.plot(X,ridge_regressor_model_2.predict(X),color='orange',label='alpha=10') # then plot the ridge regression line with alpha=10
plt.plot(X,ridge_regressor_model_3.predict(X),color='purple',label='alpha=100')# then plot the ridge regression line with alpha=100
plt.legend()
```
