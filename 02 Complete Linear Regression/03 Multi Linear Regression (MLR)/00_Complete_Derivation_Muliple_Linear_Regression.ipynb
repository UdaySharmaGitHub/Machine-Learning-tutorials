# Multiple Linear Regression (MLR)

Multiple Linear Regression extends simple linear regression to model relationships between **multiple independent variables** and a single dependent variable.

---

## Mathematical Foundation

### Model Equation:
$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$$

Or in Form of Sigma
$$\hat{y} = \beta_0 + \sum \beta_i x_i$$

Or in matrix form:
$$\hat{y} = X\beta$$


Where:
- $\hat{y}$ = predicted value
- $\beta_0$ = intercept (bias term)
- $\beta_1, \beta_2, ..., \beta_n$ = coefficients for each feature
- $x_1, x_2, ..., x_n$ = feature values

---

## Matrix Representation

### Design Matrix $X$:
For $n$ samples and $p$ features (with bias column):

$$X = \begin{bmatrix} 1 & x_{11} & x_{12} & \cdots & x_{1p} \\ 1 & x_{21} & x_{22} & \cdots & x_{2p} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 1 & x_{n1} & x_{n2} & \cdots & x_{np} \end{bmatrix}_{n \times (p+1)}$$

### Coefficient Vector $\beta$:
$$\beta = \begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \\ \vdots \\ \beta_p \end{bmatrix}_{(p+1) \times 1}$$

### Target Vector $y$:
$$y = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix}_{n \times 1}$$

### Prediction:
$$\hat{y} = X\beta$$

### Expanded Matrix Multiplication:

$$\hat{y} = X\beta = \begin{bmatrix} 1 & x_{11} & x_{12} & \cdots & x_{1p} \\ 1 & x_{21} & x_{22} & \cdots & x_{2p} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 1 & x_{n1} & x_{n2} & \cdots & x_{np} \end{bmatrix} \begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \\ \vdots \\ \beta_p \end{bmatrix}$$

Performing the matrix multiplication (each row of $X$ multiplied by $\beta$):

$$\hat{y} = \begin{bmatrix} 1 \cdot \beta_0 + x_{11} \cdot \beta_1 + x_{12} \cdot \beta_2 + \cdots + x_{1p} \cdot \beta_p \\ 1 \cdot \beta_0 + x_{21} \cdot \beta_1 + x_{22} \cdot \beta_2 + \cdots + x_{2p} \cdot \beta_p \\ \vdots \\ 1 \cdot \beta_0 + x_{n1} \cdot \beta_1 + x_{n2} \cdot \beta_2 + \cdots + x_{np} \cdot \beta_p \end{bmatrix}$$

Simplifying:

$$\hat{y} = \begin{bmatrix} \hat{y}_1 \\ \hat{y}_2 \\ \vdots \\ \hat{y}_n \end{bmatrix} = \begin{bmatrix} \beta_0 + \beta_1 x_{11} + \beta_2 x_{12} + \cdots + \beta_p x_{1p} \\ \beta_0 + \beta_1 x_{21} + \beta_2 x_{22} + \cdots + \beta_p x_{2p} \\ \vdots \\ \beta_0 + \beta_1 x_{n1} + \beta_2 x_{n2} + \cdots + \beta_p x_{np} \end{bmatrix}$$

**Each element $\hat{y}_i$ represents the prediction for the $i$-th sample:**
$$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij}$$

---

## Derivation of the Normal Equation

### Step 1: Define the Cost Function (Sum of Squared Residuals)

The residual for each observation is $e_i = y_i - \hat{y}_i$

In matrix form, the residual vector is:
$$e = y - X\beta$$

The cost function (SSR) is:
$$J(\beta) = \sum_{i=1}^{n} e_i^2 = e^T e = (y - X\beta)^T(y - X\beta)$$

### Step 2: Expand the Cost Function

$$J(\beta) = (y - X\beta)^T(y - X\beta)$$

Using the property $(A - B)^T = A^T - B^T$:
$$J(\beta) = (y^T - \beta^T X^T)(y - X\beta)$$

Expanding:
$$J(\beta) = y^T y - y^T X\beta - \beta^T X^T y + \beta^T X^T X \beta$$

Since $y^T X\beta$ is a scalar, $(y^T X\beta)^T = \beta^T X^T y = y^T X\beta$:
$$J(\beta) = y^T y - 2\beta^T X^T y + \beta^T X^T X \beta$$

### Step 3: Take the Derivative with Respect to $\beta$

Using matrix calculus rules:
- $\frac{\partial}{\partial \beta}(\beta^T X^T y) = X^T y$
- $\frac{\partial}{\partial \beta}(\beta^T X^T X \beta) = 2X^T X \beta$

$$\frac{\partial J}{\partial \beta} = -2X^T y + 2X^T X \beta$$

### Step 4: Set Derivative to Zero and Solve

$$-2X^T y + 2X^T X \beta = 0$$

$$X^T X \beta = X^T y$$

### Step 5: Solve for $\beta$ (Normal Equation) (***Final Formula***)

$$\boxed{\beta = (X^T X)^{-1} X^T y}$$

This is the **Normal Equation** - the closed-form solution for linear regression coefficients.

---

### Key Matrix Calculus Rules Used:

| Derivative | Result |
|------------|--------|
| $\frac{\partial}{\partial x}(a^T x)$ | $a$ |
| $\frac{\partial}{\partial x}(x^T a)$ | $a$ |
| $\frac{\partial}{\partial x}(x^T A x)$ | $(A + A^T)x = 2Ax$ (if $A$ is symmetric) |
