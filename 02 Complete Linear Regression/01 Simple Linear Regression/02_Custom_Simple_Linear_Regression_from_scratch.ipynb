# Simple Linear Regression using Ordinary Least Squares (OLS)

## Introduction

**Linear Regression** is a supervised machine learning algorithm used to predict a continuous target variable based on one or more input features. In **Simple Linear Regression**, we have one independent variable (X) and one dependent variable (y).

The relationship is modeled as:

$$\hat{y} = mx + b$$

Where:
- $\hat{y}$ = Predicted value
- $m$ = Slope (coefficient)
- $b$ = Intercept (bias term)
- $x$ = Input feature

---

## What is Ordinary Least Squares (OLS)?

**Ordinary Least Squares (OLS)** is a method to estimate the parameters (slope and intercept) of a linear regression model by **minimizing the sum of squared residuals (errors)** between the actual values and predicted values.

### The Cost Function

The **Sum of Squared Errors (SSE)** or **Residual Sum of Squares (RSS)** is defined as:

$$J(m, b) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{n} (y_i - (mx_i + b))^2$$

Our goal is to find values of $m$ and $b$ that **minimize** this cost function.

---

## Complete Derivation of Slope (m) and Intercept (b)

### Step 1: Define the Error Function

Let the error (residual) for each data point be:

$$e_i = y_i - \hat{y}_i = y_i - (mx_i + b)$$

The total squared error is:

$$J(m, b) = \sum_{i=1}^{n} e_i^2 = \sum_{i=1}^{n} (y_i - mx_i - b)^2$$

---

### Step 2: Find Optimal Parameters Using Calculus

To minimize $J(m, b)$, we take **partial derivatives** with respect to $m$ and $b$, and set them equal to zero.

#### Partial Derivative with respect to $b$ (Intercept):

$$\frac{\partial J}{\partial b} = \sum_{i=1}^{n} 2(y_i - mx_i - b)(-1) = 0$$

$$\sum_{i=1}^{n} (y_i - mx_i - b) = 0$$

$$\sum_{i=1}^{n} y_i - m\sum_{i=1}^{n} x_i - nb = 0$$

Dividing by $n$:

$$\bar{y} - m\bar{x} - b = 0$$

**Therefore, the intercept formula is:**

$$\boxed{b = \bar{y} - m\bar{x}}$$

Where $\bar{x}$ and $\bar{y}$ are the means of $X$ and $y$ respectively.

---

#### Partial Derivative with respect to $m$ (Slope):

$$\frac{\partial J}{\partial m} = \sum_{i=1}^{n} 2(y_i - mx_i - b)(-x_i) = 0$$

$$\sum_{i=1}^{n} x_i(y_i - mx_i - b) = 0$$

$$\sum_{i=1}^{n} x_i y_i - m\sum_{i=1}^{n} x_i^2 - b\sum_{i=1}^{n} x_i = 0$$

Substituting $b = \bar{y} - m\bar{x}$:

$$\sum_{i=1}^{n} x_i y_i - m\sum_{i=1}^{n} x_i^2 - (\bar{y} - m\bar{x})\sum_{i=1}^{n} x_i = 0$$

$$\sum_{i=1}^{n} x_i y_i - m\sum_{i=1}^{n} x_i^2 - \bar{y}\sum_{i=1}^{n} x_i + m\bar{x}\sum_{i=1}^{n} x_i = 0$$

Since $\sum_{i=1}^{n} x_i = n\bar{x}$:

$$\sum_{i=1}^{n} x_i y_i - m\sum_{i=1}^{n} x_i^2 - n\bar{x}\bar{y} + mn\bar{x}^2 = 0$$

$$\sum_{i=1}^{n} x_i y_i - n\bar{x}\bar{y} = m\left(\sum_{i=1}^{n} x_i^2 - n\bar{x}^2\right)$$

---

### Step 3: Final Formula for Slope (m)

$$\boxed{m = \frac{\sum_{i=1}^{n} x_i y_i - n\bar{x}\bar{y}}{\sum_{i=1}^{n} x_i^2 - n\bar{x}^2}}$$

This can also be written as:

$$\boxed{m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}}$$

**Interpretation:**
- **Numerator**: Covariance between X and y (measures how X and y vary together)
- **Denominator**: Variance of X (measures how X varies from its mean)

---

## Summary of OLS Formulas

| Parameter | Formula |
|-----------|---------|
| **Slope (m) or coef** | $m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}$ |
| **Intercept (b)** | $b = \bar{y} - m\bar{x}$ |

---

## Implementation of Custom Linear Regression Class

Below is our custom implementation of Simple Linear Regression using the OLS formulas derived above:

```python
class MeraLR:
    
    def __init__(self):
        self.m = None
        self.b = None
        
        # Custom Formula to Implement Linear Regression 
    def fit(self,X_train,y_train):
        
        num = 0
        den = 0
        
        for i in range(X_train.shape[0]):  # Iterates through ALL training points
            # Numerator: Sum of (x - x̄)(y - ȳ)
            num = num + ((X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean()))
            # Denominator: Sum of (x - x̄)²
            den = den + ((X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean()))
        
        # Slope or coefficient
        self.m = num/den
        # Intercept
        self.b = y_train.mean() - (self.m * X_train.mean())
        print(f"Slope (m): {self.m}")
        print(f"Intercept (b): {self.b}")       
    
    # Custom Formula to Predict the Output
    def predict(self,X_test):
        print(X_test)
        return self.m * X_test + self.b
```

### Code Explanation: `MeraLR` Class

The `MeraLR` class above implements the OLS formulas:

1. **`__init__`**: Initializes slope (`m`) and intercept (`b`) as `None`

2. **`fit(X_train, y_train)`**: Calculates slope and intercept using:
   - **Numerator (num)**: $\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})$
   - **Denominator (den)**: $\sum_{i=1}^{n} (x_i - \bar{x})^2$
   - **Slope**: $m = \frac{num}{den}$
   - **Intercept**: $b = \bar{y} - m\bar{x}$

3. **`predict(X_test)`**: Uses the learned parameters to predict: $\hat{y} = mx + b$

## Conclusion

### Key Takeaways:

1. **Ordinary Least Squares (OLS)** minimizes the sum of squared errors to find the best-fit line

2. **Slope Formula**: 
   $$m = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2} = \frac{Cov(X, y)}{Var(X)}$$

3. **Intercept Formula**: 
   $$b = \bar{y} - m\bar{x}$$

4. The regression line always passes through the point $(\bar{x}, \bar{y})$

5. **Interpretation of Slope**: For every 1 unit increase in X, y increases by $m$ units

### Advantages of OLS:
- Simple and easy to interpret
- Provides closed-form solution (no iterative optimization needed)
- Computationally efficient
- Works well when assumptions are met (linearity, homoscedasticity, independence)

### Limitations:
- Sensitive to outliers
- Assumes linear relationship between variables
- May not work well with multicollinearity in multiple regression
